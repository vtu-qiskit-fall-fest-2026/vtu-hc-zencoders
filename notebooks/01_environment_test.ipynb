# 01 - Q-Nav Simulation Environment Walkthrough
This notebook demonstrates loading 2D airspace maps, placing static obstacles, initializing the autonomous UAV drone and target, and stepping through the collision detection engine.


```python
import sys
from pathlib import Path
# Add project root to sys.path
PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import numpy as np
import matplotlib.pyplot as plt
from utils.io_utils import load_json
from simulation.environment import UAVEnvironment
from dashboard.visualization import plot_map_base
print("Environment dependencies loaded successfully.")
```

## 1. Inspecting Map and Obstacle Data
Let's load `map_01.json` and examine its obstacles and dimensions:


```python
map_data = load_json(PROJECT_ROOT / "data" / "maps" / "map_01.json")
print(f"Map ID: {map_data['map_id']}")
print(f"Dimensions: {map_data['dimensions']}")
print(f"Obstacles count: {len(map_data['obstacles'])}")
for obs in map_data['obstacles'][:3]:
    print(" ", obs)
```

## 2. Visualizing Airspace Map with Start and Target Waypoints


```python
start_coord = (5.0, 5.0)
target_coord = (45.0, 45.0)

fig, ax = plot_map_base(map_data, start=start_coord, target=target_coord)
plt.title(f"Airspace Map: {map_data['name']}", color='white', pad=12)
plt.show()
```

## 3. Initializing Environment and Stepping the Drone


```python
scenario_data = {
    "start": [5.0, 5.0],
    "target": [45.0, 45.0],
    "max_steps": 50,
}

env = UAVEnvironment(map_data=map_data, scenario_data=scenario_data)
state = env.reset()
print("Initial Drone Position:", state["drone_position"])
print("Target Distance:", state["target_distance"])
print("Nearest Obstacle Distance:", state["nearest_obstacle_distance"])

# Execute a step toward the target (direction [1.0, 1.0])
next_state, reward, done, info = env.step(direction=np.array([1.0, 1.0]))
print("\nAfter 1 step:")
print("New Position:", next_state["drone_position"])
print("Reward:", reward)
print("Done flag:", done)
```
